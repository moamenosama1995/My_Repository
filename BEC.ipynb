```python
import pandas as pd
import numpy as np
from pathlib import Path
import shutil
import os
pd.set_option("display.max_columns", 50)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

```python
PROJECT_PATH = Path("/content/BEC")
RAW_DATA_PATH = PROJECT_PATH / "data" / "raw"
CLEAN_DATA_PATH = PROJECT_PATH / "data" / "clean"

CLEAN_DATA_PATH.mkdir(parents=True, exist_ok=True)

print("Project Path :", PROJECT_PATH)
print("Raw Data     :", RAW_DATA_PATH)
print("Clean Data   :", CLEAN_DATA_PATH)
```

```python
RAW_DATA_PATH.mkdir(parents=True, exist_ok=True)

for file_name in os.listdir(CONTENT_PATH):
    if file_name.endswith('.csv'):
        shutil.move(CONTENT_PATH / file_name, RAW_DATA_PATH / file_name)
        print(f"Moved {file_name} to {RAW_DATA_PATH}")
```

```python
raw_files = sorted([file.name for file in RAW_DATA_PATH.glob("*.csv")])

print("Raw CSV files:")
for file_name in raw_files:
    print("-", file_name)

print(f"\nNumber of CSV files: {len(raw_files)}")
```

```python
Customers = pd.read_csv(
    RAW_DATA_PATH / "Customers.csv",
    sep=';',
    encoding='utf-8-sig'
)
Items = pd.read_csv(
    RAW_DATA_PATH / "Items.csv",
    sep=';',
    encoding='utf-8-sig'
)
Locations = pd.read_csv(
    RAW_DATA_PATH / "Locations.csv",
    sep=';',
    encoding='utf-8-sig'
)
Orders = pd.read_csv(
    RAW_DATA_PATH / "Orders.csv",
    sep=';',
    encoding='utf-8-sig'
)
Payments = pd.read_csv(
    RAW_DATA_PATH / "Payments.csv",
    sep=';',
    encoding='utf-8-sig'
)
Products = pd.read_csv(
    RAW_DATA_PATH / "Products.csv",
    sep=';',
    encoding='utf-8-sig'
)
Reviews = pd.read_csv(
    RAW_DATA_PATH / "Reviews.csv",
    sep=';',
    encoding='utf-8-sig'
)
Sellers = pd.read_csv(
    RAW_DATA_PATH / "Sellers.csv",
    sep=';',
    encoding='utf-8-sig'
)
```

```python
Tables = {
    "Customers": Customers,
    "Items": Items,
    "Locations": Locations,
    "Orders": Orders,
    "Payments": Payments,
    "Products": Products,
    "Reviews": Reviews,
    "Sellers": Sellers
}

print("Tables dictionary created.")
```

```python
for table_name, df in Tables.items():
    print("=" * 80)
    print("=" * 80)
    print(f"TABLE: {table_name.upper()}")
    print(f"Shape: {df.shape}")
    display(df.head(3))
```

```python
print("Converting datatypes...")

Tables['Items']['Item_ID'] = Tables['Items']['Item_ID'].astype(str)
Tables['Items']['Shipping_Limit_Date'] = pd.to_datetime(Tables['Items']['Shipping_Limit_Date'], format='%d/%m/%Y %H:%M', errors='coerce')


date_cols_orders = [
    'Order_Purchase_Date',
    'Order_Aprroved_Date',
    'Order_Delivered_Seller_Date',
    'Order_Delivered_Customer_Date',
    'Order_Estimated_Delivery_Date'
]
for col in date_cols_orders:
    Tables['Orders'][col] = pd.to_datetime(Tables['Orders'][col], format='%d/%m/%Y %H:%M', errors='coerce')

date_cols_reviews = [
    'Review_Creation_Date',
    'Review_Answer_Date'
]
for col in date_cols_reviews:
    Tables['Reviews'][col] = pd.to_datetime(Tables['Reviews'][col], format='%d/%m/%Y %H:%M', errors='coerce')

print("Datatype conversion complete.")

print("\nVerifying updated dtypes:")
for table_name in ['Items', 'Orders', 'Reviews']:
    df = Tables[table_name]
    print("=" * 80)
    print(f"TABLE: {table_name.upper()}")
    print(df.dtypes)
    print()
```

```python
for table_name, df in Tables.items():
    print("=" * 80)
    print(f"TABLE: {table_name.upper()}")
    print(df.dtypes)
    print()
```

```python
primary_keys = {
    "Customers": "Customer_ID",
    "Items": "Order_ID",
    "Orders": "Order_ID",
    "Payments": "Order_ID",
    "Locations": "Location_Zip_Code_Prefix",
    "Reviews": "Order_ID",
    "Sellers": "Seller_ID",
    "Products": "Product_ID"

}

pk_results = []

for table_name, key_column in primary_keys.items():
    df = Tables[table_name]

    pk_results.append({
        "table_name": table_name,
        "primary_key": key_column,
        "rows": len(df),
        "unique_values": df[key_column].nunique(),
        "duplicate_rows": df[key_column].duplicated().sum(),
        "missing_keys": df[key_column].isna().sum()
    })

pk_results = pd.DataFrame(pk_results)

pk_results
```

```python
print("Removing duplicate primary key rows...")

for table_name, key_column in primary_keys.items():
    df = Tables[table_name]
    initial_rows = len(df)

    if df[key_column].duplicated().any():
        Tables[table_name] = df.drop_duplicates(subset=[key_column], keep='first')
        print(f"Removed {initial_rows - len(Tables[table_name])} duplicate rows from {table_name}.")
    else:
        print(f"No duplicate primary keys found in {table_name}.")

print("Duplicate removal complete.")
```

```python
pk_results = []

for table_name, key_column in primary_keys.items():
    df = Tables[table_name]

    pk_results.append({
        "table_name": table_name,
        "primary_key": key_column,
        "rows": len(df),
        "unique_values": df[key_column].nunique(),
        "duplicate_rows": df[key_column].duplicated().sum(),
        "missing_keys": df[key_column].isna().sum()
    })

pk_results = pd.DataFrame(pk_results)

pk_results
```

```python
Customers_clean = Customers.copy()
Items_clean = Items.copy()
Locations_clean = Locations.copy()
Orders_clean = Orders.copy()
Payments_clean = Payments.copy()
Products_clean = Products.copy()
Reviews_clean = Reviews.copy()
Sellers_clean = Sellers.copy()
```

```python
print("Min transaction datetime:",
      Orders_clean["Order_Purchase_Date"].min())

print("Max transaction datetime:",
      Orders_clean["Order_Delivered_Customer_Date"].max())
```

```python
for table_name, key_column in primary_keys.items():
    df = Tables[table_name]

    duplicates = df[key_column].duplicated().sum()
    missing = df[key_column].isna().sum()

    print(
        f"{table_name:<20} | "
        f"Duplicates: {duplicates:<5} | "
        f"Missing PK: {missing}"
    )
```

```python
Customers_clean.to_csv(
    CLEAN_DATA_PATH / "Customers_clean.csv",
    index=False
)

Items_clean.to_csv(
    CLEAN_DATA_PATH / "Items_clean.csv",
    index=False
)

Orders_clean.to_csv(
    CLEAN_DATA_PATH / "Orders_clean.csv",
    index=False
)

Payments_clean.to_csv(
    CLEAN_DATA_PATH / "Payments_clean.csv",
    index=False
)

Locations_clean.to_csv(
    CLEAN_DATA_PATH / "Locations_clean.csv",
    index=False
)

Products_clean.to_csv(
    CLEAN_DATA_PATH / "Products_clean.csv",
    index=False
)

Reviews_clean.to_csv(
    CLEAN_DATA_PATH / "Reviews_clean.csv",
    index=False
)

Sellers_clean.to_csv(
    CLEAN_DATA_PATH / "Sellers_clean.csv",
    index=False
)
print("All clean tables exported successfully.")
```
